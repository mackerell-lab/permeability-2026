```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
def plot_z_scan(keywords, pdb_suffix, membrane, colors=None):
    if colors is None:
        colors = {
            'benc': 'purple',
            'prpc': 'green',
            'dmeo': 'red',
            'imin': 'magenta',
            'meoo': 'brown',
            'foro': 'blue',
            'mamn': 'cyan',
            'acec': 'orange',
            'tipo': 'black',
            'clex': 'blue',
            'clbx': 'green',
            'brbx': 'red',
            'tfec': 'brown',
            'flbx': 'gray',
            'flbc': 'cyan',
            'fetx': 'pink'
        }

    datadir = "../data"
    for scheme in ["big_voxel", "default_unit_voxel"]:
        voxel_scheme = scheme + "_scheme"
        scheme_title = scheme.replace("_", " ")
        plt.figure(figsize=(4, 3))
        # Loop through suffixes and process each file
        for suffix in keywords:
            # Construct the filename
            fname = 'fragmaps/' + voxel_scheme + '/maps/' + str(pdb_suffix) + '.' + str(suffix) + '.gfe_map.npy'
            ylabel = "GFE (kcal/mol)"
        
            filename = os.path.join(datadir, membrane, fname)
            
            try:
                # Load the matrix
                matrix = np.load(filename)
                midpoint = len(matrix) // 2
                lower = matrix[:(midpoint + 1)]
                upper = matrix[midpoint:]
                reverse_lower = lower[::-1]
                avg_matrix = (reverse_lower+upper)/2
                print(f"Loaded matrix from {filename} with shape {matrix.shape}")
                
                # Plot the matrix
                # plt.plot(matrix, marker='o', label=suffix)
                # Plot a horizontal line at y=0
                plt.axhline(y=0, color='grey', linestyle='--', linewidth=2)
                # plt.plot(avg_matrix, marker='o', label=suffix)
                plt.plot(avg_matrix[:-10], label=suffix, color=colors.get(suffix, 'black'))
            except FileNotFoundError:
                print(f"File {filename} not found. Skipping...")

        # plt.title(f"fragment GFE ")
        plt.legend(loc="upper right")
        # plt.title(scheme_title)
        plt.xlabel("Distance from Bilayer Center " + r'($\AA$)')
        plt.ylabel(ylabel)
        # plt.tight_layout()
        plt.ylim([-2,6])
        filename = [datadir, membrane, voxel_scheme]
        filename.extend(keywords)
        plt.savefig(
            "_".join(filename) + ".png",
            dpi=300,
            bbox_inches="tight",
            pad_inches=0.02
        )
        plt.show()




# List of keywords for fragments
keywords_full = ['acec', 'apolar', 'benc', 'dmeo', 'forc', 'forn', 'foro', 'gehc', 'hbacc', 'hbdon', 'imin', 'iminh', 'mamn', 'meoo', 'prpc', 'tipo'] 
keywords_0 = ['benc', 'prpc', 'dmeo'] 
keywords_1 = ['imin', 'meoo', 'foro']
keywords_2 = ['mamn', 'acec']
keywords_3 = ['imin', 'meoo', 'foro',  'mamn', 'acec']
keywords_4 = ['benc', 'prpc', 'dmeo', 'imin', 'meoo', 'foro',  'mamn', 'acec']
keyword_sol = ['tipo']
keywords_halo=['clex', 'clbx', 'brbx', 'tfec', 'flbx', 'fetx'] # 'flbc' 
```

```python
membrane = "POPC-CHOL"
pdb_suffix = "popc_chol_dummy"


plot_z_scan(keywords_0,pdb_suffix, membrane)
plot_z_scan(keywords_1,pdb_suffix, membrane)
plot_z_scan(keywords_2,pdb_suffix, membrane)
plot_z_scan(keywords_3,pdb_suffix, membrane)
plot_z_scan(keywords_4,pdb_suffix, membrane)
plot_z_scan(keyword_sol,pdb_suffix, membrane)
# plot_gfe_z_scan(keywords_3,pdb_suffix, datadir, membrane, voxel_scheme, slab_size)
```
